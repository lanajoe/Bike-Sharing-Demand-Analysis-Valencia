### Preprocesamiento de datos 

Este apartado nos a ayuda a seleccionar las variables que permiten trabajar con un conjunto de datos más sencillo y adecuado para el desarrollo del modelo predictivo. 

```python
import pandas as pd

# 1. Carga del archivo usando la ruta absoluta de tu Mac
df = pd.read_csv("/Users/mac/documents/proyecto/data/interim/usobarriosmeteo.csv", sep=";")

# 2. Creación de la columna de horas usando la operación módulo
df['hora_dia'] = df['hr'] % 24

# 3. Creación de la columna fin de semana a partir de la fecha
# Convertimos la columna 'fecha' a formato fecha de Pandas
fechas_transformadas = pd.to_datetime(df['fecha'])
# .dt.weekday devuelve un número del 0 (lunes) al 6 (domingo). Sábado(5) y Domingo(6) son > 4.
df['es_fin_semana'] = (fechas_transformadas.dt.weekday > 4).astype(int)

# 4. Filtrar los registros para quedarnos SOLO con tu barrio de estudio (LA SEU)
df_seu = df[df['name'] == 'LA SEU']

# 5. Seleccionar las columnas finales que pide el enunciado para el modelo
columnas_finales = ['hora_dia', 'es_fin_semana', 'ta', 'prec', 'uso_bici']
df_processed = df_seu[columnas_finales]

# 6. Guardar el resultado en la carpeta 'processed' en formato CSV
df_processed.to_csv("/Users/mac/documents/proyecto/data/processed/usobarriosmeteo_processed.csv", sep=";", index=False)

# 7. Comprobación del resultado en pantalla
print("Tamaño del nuevo dataset (filas, columnas):", df_processed.shape)
display(df_processed.head())
```
