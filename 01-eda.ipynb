### Análisis Exploratorio de Datos (EDA)

El objetivo de este apartado es analizar cómo se comporta el flujo de usuarios en el barrio de estudio, La Seu, y compararlo con el barrio de referencia, Beniferri. Para ello, se pretende identificar posibles diferencias en los patrones horarios de uso y evaluar en qué medida las condiciones meteorológicas pueden influir en la demanda observada.

```python
# instalamos librería pandas
%pip install pandas
```

```python
import pandas as pd
import numpy as np

# 1. Carga del archivo usando la ruta absoluta completa de tu Mac
df = pd.read_csv("/Users/mac/documents/proyecto/data/interim/usobarriosmeteo.csv", sep=";")

# 2. Filtrar los datos para quedarnos solo con tus dos barrios
# Buscamos en la columna 'name' los registros que sean exactamente 'LA SEU' o 'BENIFERRI'
df_filtrado = df[(df['name'] == 'LA SEU') | (df['name'] == 'BENIFERRI')]

# 3. Crear la tabla pivote (Pivot Table)
# Filas: hora | Columnas: name | Valores a calcular: uso_bici y ta (temperatura)
# Usamos la media (np.mean) para agrupar los datos
eda_pivot = pd.pivot_table(
    df_filtrado,
    values=['uso_bici', 'ta'],
    index='hr',
    columns='name',
    aggfunc=np.mean
)

# 4. Mostrar el resultado final en pantalla redondeado a 2 decimales
print(" TABLA DE ENTRADA: HORAS, TEMPERATURA Y USO DE BICIS ")
eda_pivot.round(2)
```
