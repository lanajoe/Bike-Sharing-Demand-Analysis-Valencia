### Investigación y referencias 

El estudio de Ereni y Román (2021) analiza la influencia de las condiciones meteorológicas y las características de los barrios en el uso de los sistemas de bicicletas compartidas. Los autores concluyen que la demanda no es igual en toda la ciudad, sino que depende del tipo de zona analizada. 

Según sus resultados, los barrios con una mayor actividad comercial, administrativa o turística mantienen una demanda relativamente estable durante el día. En cambio, las áreas residenciales presentan patrones más ligados a los horarios de entrada y salida del trabajo o de los centros educativos, además de una mayor sensibilidad a las bajas temperaturas. 

A partir de las conclusiones del estudio, se plantean varias mejoras para el análisis realizado: 

Incorporar una variable que diferencie los fines de semana de los días laborables, ya que los patrones de movilidad cambian según el tipo de día. 

Analizar por separado los datos de La Seu y Beniferri, puesto que ambos barrios presentan comportamientos de demanda diferentes. 

Considerar el efecto de las temperaturas más bajas, ya que estas pueden influir en la reducción del uso del servicio en determinadas franjas horarias.
